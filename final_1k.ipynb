```python
import os
import shutil
import json

# === 1. Define source and target paths ===
source_train = "/kaggle/input/images/images/train2017"
source_val = "/kaggle/input/images/images/val2017"
source_ann_dir = "/kaggle/input/annott/annotations"

target_train = "/kaggle/working/YOLO-main/data/coco/images/train2017"
target_val = "/kaggle/working/YOLO-main/data/coco/images/val2017"
target_ann_dir = "/kaggle/working/YOLO-main/data/coco/annotations"

os.makedirs(target_train, exist_ok=True)
os.makedirs(target_val, exist_ok=True)
os.makedirs(target_ann_dir, exist_ok=True)

# === 2. Copy 1,000 training images if not already copied ===
train_files = sorted([f for f in os.listdir(source_train) if f.endswith(('.jpg', '.png'))])
selected_train_files = train_files[:1000]

copied_train = 0
for f in selected_train_files:
    src = os.path.join(source_train, f)
    dst = os.path.join(target_train, f)
    if not os.path.exists(dst):
        shutil.copy(src, dst)
        copied_train += 1
print(f"✅ Skipped existing, copied {copied_train} new training images")

# === 3. Copy 200 validation images if not already copied ===
val_files = sorted([f for f in os.listdir(source_val) if f.endswith(('.jpg', '.png'))])
selected_val_files = val_files[:200]

copied_val = 0
for f in selected_val_files:
    src = os.path.join(source_val, f)
    dst = os.path.join(target_val, f)
    if not os.path.exists(dst):
        shutil.copy(src, dst)
        copied_val += 1
print(f"✅ Skipped existing, copied {copied_val} new validation images")

# === 4. Copy annotation files if not already present ===
for ann_file in ['instances_train2017.json', 'instances_val2017.json']:
    src = os.path.join(source_ann_dir, ann_file)
    dst = os.path.join(target_ann_dir, ann_file)
    if not os.path.exists(dst):
        shutil.copy(src, dst)
print("✅ Skipped existing, copied annotation files if missing")

# === 5. Filter TRAIN annotations to match 1,000 selected images ===
train_ann_path = os.path.join(target_ann_dir, 'instances_train2017.json')
filtered_train_ann_path = os.path.join(target_ann_dir, 'instances_train2017_1k.json')

if not os.path.exists(filtered_train_ann_path):
    with open(train_ann_path, 'r') as f:
        coco = json.load(f)

    used_image_set = set(selected_train_files)
    filtered_images = [img for img in coco['images'] if img['file_name'] in used_image_set]
    used_image_ids = {img['id'] for img in filtered_images}
    filtered_annotations = [ann for ann in coco['annotations'] if ann['image_id'] in used_image_ids]

    filtered_coco = {
        'info': coco.get('info', {}),
        'licenses': coco.get('licenses', []),
        'categories': coco['categories'],
        'images': filtered_images,
        'annotations': filtered_annotations
    }

    with open(filtered_train_ann_path, 'w') as f:
        json.dump(filtered_coco, f)
    print(f"✅ Filtered train annotation saved to: {filtered_train_ann_path}")
else:
    print(f"✅ Filtered train annotation already exists: {filtered_train_ann_path}")

# === 6. Filter VAL annotations to match 200 selected images ===
val_ann_path = os.path.join(target_ann_dir, 'instances_val2017.json')
filtered_val_ann_path = os.path.join(target_ann_dir, 'instances_val2017_200.json')

if not os.path.exists(filtered_val_ann_path):
    with open(val_ann_path, 'r') as f:
        coco = json.load(f)

    used_val_set = set(selected_val_files)
    filtered_val_images = [img for img in coco['images'] if img['file_name'] in used_val_set]
    used_val_ids = {img['id'] for img in filtered_val_images}
    filtered_val_annotations = [ann for ann in coco['annotations'] if ann['image_id'] in used_val_ids]

    filtered_val_coco = {
        'info': coco.get('info', {}),
        'licenses': coco.get('licenses', []),
        'categories': coco['categories'],
        'images': filtered_val_images,
        'annotations': filtered_val_annotations
    }

    with open(filtered_val_ann_path, 'w') as f:
        json.dump(filtered_val_coco, f)
    print(f"✅ Filtered val annotation saved to: {filtered_val_ann_path}")
else:
    print(f"✅ Filtered val annotation already exists: {filtered_val_ann_path}")
```
